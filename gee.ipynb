```python
import ee
```

```python
ee.Authenticate()
```

```python
ee.Initialize()
```

```python
roi = ee.Geometry.Rectangle([70, 20, 90, 30])  # Example: North India
```

```python
ndvi = ee.ImageCollection("MODIS/061/MOD13Q1") \
    .filterDate("2020-01-01", "2020-01-10") \
    .select("NDVI") \
    .mean()

# Scale factor correction (IMPORTANT)
ndvi = ndvi.multiply(0.0001)
```

```python
lst = ee.ImageCollection("MODIS/061/MOD11A1") \
    .filterDate("2020-01-01", "2020-01-10") \
    .select("LST_Day_1km") \
    .mean()

lst = lst.multiply(0.02)  # Kelvin scaling
```

```python
task = ee.batch.Export.image.toDrive(
    image=ndvi,
    description='NDVI_export',
    scale=500,
    region=roi,
    fileFormat='GeoTIFF'
)
task.start()
```
