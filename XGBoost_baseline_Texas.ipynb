```python
import pandas as pd
import numpy as np
from sklearn.model_selection import GroupShuffleSplit, train_test_split
from sklearn.metrics import mean_squared_error, r2_score
from xgboost import XGBRegressor
```

```python
df = pd.read_parquet("Data/Texas/Texas_full.parquet")

# basic cleanup
df = df.dropna(subset=["sm", "VV", "VH", "NDVI", "NDWI", "LST", "SMAP"])

# ensure correct types
df["timestamp"] = pd.to_datetime(df["timestamp"])
```

```python
df["doy"] = df["timestamp"].dt.dayofyear
df["hour"] = df["timestamp"].dt.hour

df["VV_VH_ratio"] = df["VV"] / (df["VH"] + 1e-6)
df["VV_minus_VH"] = df["VV"] - df["VH"]

# optional: cyclic encoding
df["doy_sin"] = np.sin(2 * np.pi * df["doy"] / 365)
df["doy_cos"] = np.cos(2 * np.pi * df["doy"] / 365)
df["station_code"] = df["point_id"].astype("category").cat.codes

df['precip_log'] = np.log1p(df['precip'])
df['ssr_log'] = np.log1p(df['ssr'])

df['ERA5_SM_norm'] = df.groupby('point_id')['ERA5_SM'] \
    .transform(lambda x: (x - x.mean()) / x.std())
```

```python
print(df["ERA5_SM"].describe())
print(df.groupby("point_id")["ERA5_SM"].std().describe())
print(df[["ERA5_SM", "sm"]].corr())
```

```python
df["target"] = df["sm"] - df["ERA5_SM"]
target = "target"

features = [
    "VV", "VH",
    "NDVI", "NDWI", "LST", "LAI",
    "ERA5_SM", "precip_log", "ssr_log",
    "SMAP_norm",
    "clay_frac", "sand_frac", "silt_frac",
    "doy_sin", "doy_cos"
]

df = df.dropna(subset=features)

X = df[features]
y = df[target]
groups = df["point_id"]
```

```python
from sklearn.model_selection import GroupShuffleSplit

gss = GroupShuffleSplit(n_splits=1, test_size=0.2, random_state=42)

train_idx, test_idx = next(gss.split(df, groups=df["point_id"]))

train_df = df.iloc[train_idx].reset_index(drop=True)
test_df  = df.iloc[test_idx].reset_index(drop=True)

X_train = train_df[features]
y_train = train_df[target]

X_test = test_df[features]
y_test = test_df[target]
groups_train = train_df["point_id"]
```

```python
model = XGBRegressor(
    n_estimators=500,
    learning_rate=0.05,
    max_depth=6,
    subsample=0.8,
    colsample_bytree=0.8,
    reg_lambda=1.0,
    objective="reg:squarederror",
    tree_method="hist",
    random_state=42
)
```

```python
model.fit(X_train, y_train)
```

```python
pred = model.predict(X_test)
sm_true = test_df["sm"]
sm_pred = pred + test_df["ERA5_SM"]

rmse = np.sqrt(mean_squared_error(sm_true, sm_pred))
r2 = r2_score(sm_true, sm_pred)
```

```python
importances = pd.Series(model.feature_importances_, index=features)
importances = importances.sort_values(ascending=False)
print("\n===== XGBOOST BASELINE =====")
print(f"RMSE: {rmse:.5f}")
print(f"R2  : {r2:.5f}")

print("\n===== FEATURE IMPORTANCE =====")
print(importances)
```
